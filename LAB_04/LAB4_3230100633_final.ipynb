### add Gaussian Noise and Salt-and-Pepper Noise.

```python
def imageNormalization(img, dtype: np.dtype = np.float64):
    # normalization the image to [0, 255]
    img_normalized = (img - img.min()) / (img.max() - img.min())
    return img_normalized.astype(dtype)

def addGaussianNoise(img, mean: float = 0, var: float = 0.01):
    """ add Gaussian noise to the image """
    sigma = np.sqrt(var)
    noise = np.random.normal(mean, sigma, img.shape)
    noisy_img = img + noise
    return noisy_img

def addSaltAndPepperNoise(img, salt_prob: float = 0.01, pepper_prob: float = 0.01, dtype: np.dtype = np.float64):
    """ add salt and pepper noise to the image """
    img = imageNormalization(img, dtype=dtype)
    noisy_img = img.copy()

    # calculate the number of salt and pepper noise
    num_salt = np.ceil(salt_prob * img.size)
    num_pepper = np.ceil(pepper_prob * img.size)

    # add salt noise
    salt_coords = [np.random.randint(0, i - 1, int(num_salt)) for i in img.shape]
    noisy_img[tuple(salt_coords)] = 1

    # add pepper noise
    pepper_coords = [np.random.randint(0, i - 1, int(num_pepper)) for i in img.shape]
    noisy_img[tuple(pepper_coords)] = 0

    return noisy_img.astype(dtype)

# test
gauss_mean = 0
gauss_var = 0.02
img_add_gauss = addGaussianNoise(original_img, mean=gauss_mean, var=gauss_var)

sp_salt_p = 0.03
sp_pep_p = 0.03
img_add_sp = addSaltAndPepperNoise(original_img, salt_prob=sp_salt_p, pepper_prob=sp_pep_p)

# show the infromation of the image
print("Name: Add Gaussian Noise")
print("Shape:", img_add_gauss.shape)
print("Data type:", img_add_gauss.dtype)
print("MAX value:", img_add_gauss.max())
print("MIN value:", img_add_gauss.min())

print()
print("Name: Add Salt and Pepper Noise")
print("Shape:", img_add_sp.shape)
print("Data type:", img_add_sp.dtype)
print("MAX value:", img_add_sp.max())
print("MIN value:", img_add_sp.min())

# show the image
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(imageNormalization(original_img), cmap='gray')
axes[0].set_title('Original Image')
axes[1].imshow(imageNormalization(img_add_gauss), cmap='gray')
axes[1].set_title(f'Gaussian Noise\n(mean={gauss_mean}, var={gauss_var})')
axes[2].imshow(imageNormalization(img_add_sp), cmap='gray')
axes[2].set_title(f'S&P Noise\n(salt={sp_salt_p}, pepper={sp_pep_p})')
for ax in axes: 
    ax.axis('off')
plt.show()
```

```python
def quantify_histogram(img, dtype: np.dtype = np.float64):
    """
    quantify the image according to the histogram
    """

    # change the image to dtype
    image = img.astype(dtype)
    image_min = image.min()
    image_max = image.max()

    if image_max == image_min:
        output_image = np.zeros(256, dtype=dtype)
        output_image[0] = 1.0
        return output_image

    # normalize the image 
    image_norm = imageNormalization(image, dtype=dtype)
    image_quantized = np.clip((image_norm * 255).astype(np.int32), 0, 255)

    output_image = np.bincount(image_quantized.ravel(), minlength=256).astype(np.float32)
    output_image = output_image / output_image.sum()
    return output_image
```

#### Estimate Gaussian Noise

```python
def estimateGaussianNoise(img_clean, img_noisy, dtype: np.dtype = np.float64):
    """ estimate Gaussian Noise """
    # calculate the noise
    noise = img_noisy - img_clean

    # calculate the mean and variance of the noise
    mean = np.mean(noise)
    var = np.var(noise)

    return mean, var

# test
gauss_mean = 100
gauss_var = 0.1
img_add_gauss = addGaussianNoise(original_img, mean=gauss_mean, var=gauss_var)
estimated_mean_gauss, estimated_var_gauss = estimateGaussianNoise(original_img, img_add_gauss)

# show the result
print(f"Actual Gaussian Noise - Mean: {gauss_mean:.4f}, Variance: {gauss_var:.4f}")
print(f"Estimated Gaussian Noise - Mean: {estimated_mean_gauss:.4f}, Variance: {estimated_var_gauss:.4f}")
```

The estimated results are very close to the actual values.

#### Estimate Salt-and-Pepper Noise

```python
def estimateSPNoise(img_clean, img_noisy, dtype: np.dtype = np.float64):
    """ estimate Salt and Pepper Noise """
    # calculate the noise
    img_clean_norm = imageNormalization(img_clean, dtype=dtype)
    img_noisy_norm = imageNormalization(img_noisy, dtype=dtype)

    # calculate the salt and pepper noise mask
    changed = img_clean_norm != img_noisy_norm
    salt_mask = changed & (img_noisy_norm == 1)
    pepper_mask = changed & (img_noisy_norm == 0)
    total = img_clean.size

    # calculate the salt and pepper noise probability
    salt_prob = np.sum(salt_mask) / total
    pepper_prob = np.sum(pepper_mask) / total

    return salt_prob, pepper_prob

# test 
sp_salt_p = 0.03
sp_pep_p = 0.03
img_add_sp = addSaltAndPepperNoise(original_img, salt_prob=sp_salt_p, pepper_prob=sp_pep_p)
estimated_salt_prob, estimated_pepper_prob = estimateSPNoise(original_img, img_add_sp)

# show the result
print(f"Actual S&P Noise - Salt Probability: {sp_salt_p:.4f}, Pepper Probability: {sp_pep_p:.4f}")
print(f"Estimated S&P Noise - Salt Probability: {estimated_salt_prob:.4f}, Pepper Probability: {estimated_pepper_prob:.4f}")
```

```python
# quantify: histogram
def quantify_histogram(img, dtype: np.dtype = np.float64):
    """
    quantify the image according to the histogram
    """

    # change the image to dtype
    img = img.astype(dtype)
    image_min = img.min()
    image_max = img.max()

    if image_max == image_min:
        output_image = np.zeros(256, dtype=dtype)
        output_image[0] = 1.0
        return output_image

    # normalize the image 
    image_norm = imageNormalization(img, dtype=dtype)
    image_quantized = np.clip((image_norm * 255).astype(np.int32), 0, 255)

    output_image = np.bincount(image_quantized.ravel(), minlength=256).astype(np.float32)
    output_image = output_image / output_image.sum()
    return output_image

# quantify: psnr
def quantify_psnr(img_original, img_fixed, max_pixel: float = 255.0, dtype: np.dtype = np.float64):
    """ quantify the image quality using PSNR """
    # calculate the mean squared error
    mse = np.mean((img_original.astype(dtype) - img_fixed.astype(dtype)) ** 2)

    if mse == 0:
        return np.inf

    # calculate the PSNR
    psnr = 10 * np.log10((max_pixel ** 2) / mse)
    return psnr

# show images and histograms
def show_images(images, titles, figsize=(15, 5), cmap='gray'):
    """ show images in a row """
    fig, axes = plt.subplots(2, len(images), figsize=(5 * len(images), 10))
    bins = np.arange(256)
    for i, (img, title) in enumerate(zip(images, titles)):
        # first row: images
        axes[0, i].imshow(imageNormalization(img), cmap=cmap)
        axes[0, i].set_title(title)
        axes[0, i].axis('off')

        # second row: histograms
        images_hist = quantify_histogram(img)
        axes[1, i].bar(bins, images_hist, width=1.0)
        axes[1, i].set_title(f'{title} Histogram')
        axes[1, i].set_xlim(0, 255)

    plt.tight_layout()
    plt.show()
```

### Apply morphological operations to the noisy_brain_mask.npy to denoise, remove wrong segment boundary and fill partial internal holes.

Now, have the basic binary function for **Erosion** and **Dilation**.

```python
def binaryErosion(img, kernel_size: int = 3, dtype: np.dtype = np.bool):
    """ binary erosion """
    img = img.astype(dtype)
    eroded = np.copy(img)
    pad_size = kernel_size // 2
    padded_img = np.pad(img, pad_size, mode='constant', constant_values=0)
    
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            region = padded_img[i:i+kernel_size, j:j+kernel_size]
            if np.all(region == 1):
                eroded[i, j] = 1
            else:
                eroded[i, j] = 0

    return eroded.astype(dtype)

def binaryDilation(img, kernel_size: int = 3, dtype: np.dtype = np.bool):
    """ binary dilation """
    img = img.astype(dtype)
    dilated = np.copy(img)
    pad_size = kernel_size // 2
    padded_img = np.pad(img, pad_size, mode='constant', constant_values=0)
    
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            region = padded_img[i:i+kernel_size, j:j+kernel_size]
            if np.any(region == 1):
                dilated[i, j] = 1
            else:
                dilated[i, j] = 0

    return dilated.astype(dtype)
```

```python
# erosion first
erosion_kernel_size = 3
eroded_brain_mask = binaryErosion(brain_mask, kernel_size=erosion_kernel_size)

# dilation second
dilation_kernel_size = 5
dilated_brain_mask = binaryDilation(eroded_brain_mask, kernel_size=dilation_kernel_size)

back_kernel_size = dilation_kernel_size - erosion_kernel_size + 1
fixed_brain_mask = binaryErosion(dilated_brain_mask, kernel_size=back_kernel_size)

# show the image
show_images_project2(
    images=[brain_mask, eroded_brain_mask, dilated_brain_mask, fixed_brain_mask],
    titles=[
        'Original Brain Mask', 
        f'Eroded Brain Mask\nkernel size: {erosion_kernel_size}', 
        f'Dilated Brain Mask\nkernel size: {dilation_kernel_size}',
        f'Fixed Brain Mask\nkernel size: {back_kernel_size}'
    ]
)
```

```python
def overlayMask(img_background, img_mask, color: tuple = (255, 0, 0), alpha: float = 0.5, dtype: np.dtype = np.float64):
    """ overlay the mask on the background image """
    background = np.asarray(img_background, dtype=dtype)
    mask = np.asarray(img_mask, dtype=dtype)

    if background.ndim == 2:
        background = imageNormalization(background, dtype=dtype)
        background_rgb = np.stack([background, background, background], axis=-1)
    else:
        if background.max() > 1.0:
            background = background / 255.0
        background_rgb = np.clip(background, 0, 1)
    
    mask_valid = mask > 0

    red = np.zeros_like(background_rgb, dtype=dtype)
    red[..., 0] = 1.0  # R 通道设为1

    overlay = background_rgb.copy()
    overlay[mask_valid] = (
        (1 - alpha) * background_rgb[mask_valid] +
        alpha * red[mask_valid]
    )

    overlay = np.clip(overlay * 255.0, 0, 255).astype(dtype)
    return overlay
```

### Blurring the image with a Guassian filter in the frequency domain — Set different D0

A Gaussian low-pass filter in frequency domain is

$$
H(u,v)=\frac{1}{\sqrt{2\pi}\sigma}\exp\left(-\dfrac{u^2 + v^2}{2\sigma^2}\right)
$$

The coefficient has no effect on the result, so simplified it:

$$
H(u,v)=\exp\left(-\dfrac{u^2 + v^2}{2\sigma^2}\right)
$$

```python
def gaussianLowpass(shape: tuple, sigma: float, dtype: np.dtype = np.float64):
    """ create a Gaussian lowpass filter in the frequency domain """
    rows, cols = shape
    u = np.arange(rows) - rows / 2
    v = np.arange(cols) - cols / 2
    V, U = np.meshgrid(v, u)
    D2 = U**2 + V**2
    H = np.exp(-D2 / (2 * sigma**2))
    return H.astype(dtype)

def applyGaussianNoise(img, D0: float, dtype: np.dtype = np.float64):
    """ apply a filter in the frequency domain """
    img = img.astype(dtype)
    F = np.fft.fft2(img)
    F = np.fft.fftshift(F)
    H = gaussianLowpass(img.shape, D0, dtype=dtype)
    G = F * H
    blurred = np.fft.ifft2(np.fft.ifftshift(G))
    return np.real(blurred).astype(dtype), H
```

```python
# D0 list
D0_list = [10, 30, 50, 70, 90]
blurred_results = []

# apply the filter with different D0
for D0 in D0_list:
    blurred_img, gaussian_filter = applyGaussianNoise(img, D0)
    blurred_results.append({
        "D0": D0,
        "blurred_img": blurred_img,
        "gaussian_filter": gaussian_filter
    })

# show the image
show_images_project2(
    images=[result["blurred_img"] for result in blurred_results],
    titles=[f'Blurred Image\n(D0={result["D0"]})' for result in blurred_results]
)
```

### Add Rician noise

```python
def applyRicianNoise(img, sigma: float, seed: int = None, dtype: np.dtype = np.float64):
    """ apply Rician noise to the image """
    img = img.astype(dtype)
    rng = np.random.default_rng(seed)
    noise_real = rng.normal(0, sigma, img.shape)
    noise_imag = rng.normal(0, sigma, img.shape)
    noisy_img = np.sqrt((img + noise_real)**2 + noise_imag**2)
    return noisy_img.astype(dtype)

sigma_list = [5, 10, 20, 30, 40]
noisy_results = []

# apply Rician noise with different sigma
for sigma in sigma_list:
    noisy_img = applyRicianNoise(img, sigma)
    noisy_results.append({
        "sigma": sigma,
        "noisy_img": noisy_img
    })

# show the image
show_images_project2(
    images=[result["noisy_img"] for result in noisy_results],
    titles=[f'Rician Noisy Image\n(sigma={result["sigma"]})' for result in noisy_results]
)
```

```python
def applyInverseFilter(img, H, eps: float = 1e-3, dtype: np.dtype = np.float64):
    """ apply inverse filter in the frequency domain """
    img = img.astype(dtype)
    G = np.fft.fft2(img)
    G = np.fft.fftshift(G)
    F_hat = G / np.where(np.abs(H) < eps, eps, H)
    restored = np.fft.ifft2(np.fft.ifftshift(F_hat))
    return np.real(restored).astype(dtype)

# test
D0 = 30
sigma = 10
img_add_gauss, H = applyGaussianNoise(img, D0)
img_add_ricain = applyRicianNoise(img_add_gauss, sigma)
img_add_gauss_restored = applyInverseFilter(img_add_ricain, H)

# show the image
show_images_project2(
    images=[img, img_add_gauss, img_add_gauss_restored],
    titles=['Original Image', 'Gaussian Noisy Image', 'Restored Image\n(Inverse Filter)']
)
```

The results are consistent with what was covered in the theoretical lectures: a direct inverse operation amplifies the noise, rendering the image unrecognizable.

#### Limited Inverse Filter

```python
def applyLimitedInverseFilter(img, H, R: float, eps: float = 1e-3, dtype: np.dtype = np.float64):
    """ apply limited inverse filter in the frequency domain """
    # get meshgrid of img shape
    rows, cols = img.shape
    u = np.arange(0, cols) - cols / 2
    v = np.arange(0, rows) - rows / 2
    U, V = np.meshgrid(u, v)
    D = np.sqrt(U**2 + V**2)
    mask = (D <= R).astype(dtype)

    # apply inverse filter with mask limit
    img = img.astype(dtype)
    G = np.fft.fftshift(np.fft.fft2(img))
    H_hat = G / np.where(np.abs(H) < eps, eps, H) * mask
    restored = np.fft.ifft2(np.fft.ifftshift(H_hat))
    return np.real(restored).astype(dtype)

# add noise
D0 = 30
sigma = 10
img_add_gauss, H = applyGaussianNoise(img, D0)
img_add_ricain = applyRicianNoise(img_add_gauss, sigma)

# R list
R_list = [10, 20, 30, 40, 50]
restored_results = []

# apply limited inverse filter with different R
for R in R_list:
    img_restored_limited = applyLimitedInverseFilter(img_add_ricain, H, R)
    restored_results.append({
        "R": R,
        "restored_img": img_restored_limited
    })

# show the image
show_images_project2(
    images=[result["restored_img"] for result in restored_results],
    titles=[f'Restored Image\n(Limited Inverse Filter)\nR={result["R"]}' for result in restored_results]
)
```

Reconstruction results are limited.

#### Wiener filter

```python
def applyWienerFilter(img, H, K: float = 0.01, eps: float = 1e-8, dtype: np.dtype = np.float64):
    """ apply Wiener filter to the noisy image """
    img = img.astype(dtype)
    G = np.fft.fftshift(np.fft.fft2(img))
    H_conj = np.conj(H)
    W = H_conj / (np.abs(H)**2 + K + eps)
    F_hat = W * G
    restored = np.fft.ifft2(np.fft.ifftshift(F_hat))
    return np.real(restored).astype(dtype), W

# add noise
D0 = 30
sigma = 10
img_add_gauss, H = applyGaussianNoise(img, D0)
img_add_ricain = applyRicianNoise(img_add_gauss, sigma)

# K list
K_list = [0.001, 0.01, 0.1, 1, 10]
restored_results = []

# apply Wiener filter with different K
for K in K_list:
    img_restored_wiener, W = applyWienerFilter(img_add_ricain, H, K)
    restored_results.append({
        "K": K,
        "restored_img": img_restored_wiener,
        "filter_kernel": W
    })

# show the image
show_images_project2(
    images=[result["filter_kernel"] for result in restored_results],
    titles=[f'Wiener Filter Kernel\nK={result["K"]}' for result in restored_results]
)

show_images_project2(
    images=[result["restored_img"] for result in restored_results],
    titles=[f'Restored Image\n(Wiener Filter)\nK={result["K"]}' for result in restored_results]
)
```

Selecting an appropriate value for `K` has a significant impact on the repair results.

Just try different kernel, and select the best result for the evaluations. So we must have the function to evaluate the restored image quality first.

```python
# mse
def quantify_mse(img_original, img_fixed, dtype: np.dtype = np.float64):
    """ quantify the image quality using MSE """
    img_original = img_original.astype(dtype)
    img_fixed = img_fixed.astype(dtype)
    return np.mean((img_original - img_fixed) ** 2)

# psnr
def quantify_psnr(img_original, img_fixed, data_range: float = 255.0, dtype: np.dtype = np.float64):
    """ quantify the image quality using PSNR """
    from skimage.metrics import peak_signal_noise_ratio as psnr
    img_original = img_original.astype(dtype)
    img_fixed = img_fixed.astype(dtype)
    return psnr(img_original, img_fixed, data_range=data_range)

# ssim 
def quantify_ssim(img_original, img_fixed, data_range: float = 255.0, dtype: np.dtype = np.float64):
    """ quantify the image quality using SSIM """
    from skimage.metrics import structural_similarity as ssim
    img_original = img_original.astype(dtype)
    img_fixed = img_fixed.astype(dtype)
    return ssim(img_original, img_fixed, data_range=data_range)

# evaluate the restoration results
def evaluate_image(img_original, img_restored, data_range: float = 255.0, dtype: np.dtype = np.float64):
    """ evaluate the image restoration results using MSE, PSNR, and SSIM """
    mse = quantify_mse(img_original, img_restored, dtype=dtype)
    psnr = quantify_psnr(img_original, img_restored, data_range=data_range, dtype=dtype)
    ssim = quantify_ssim(img_original, img_restored, data_range=data_range, dtype=dtype)
    return {
        "MSE": float(mse),
        "PSNR": float(psnr),
        "SSIM": float(ssim)
    }
```

```python
# test 

# K list
K_list = [0.001, 0.01, 0.1, 1, 10]
restored_results = []

# apply Wiener filter with different K
for K in K_list:
    img_restored_wiener, W = applyWienerFilter(img_add_ricain, H, K)
    restored_results.append({
        "K": K,
        "restored_img": img_restored_wiener,
        "filter_kernel": W
    })

# evaluate the restoration results
for result in restored_results:
    evaluation = evaluate_image(img, result["restored_img"])
    print(f"K={result['K']}: MSE={evaluation['MSE']:.4f}, PSNR={evaluation['PSNR']:.4f} dB, SSIM={evaluation['SSIM']:.4f}")

# show the filter 
show_images_project2(
    images=[result["filter_kernel"] for result in restored_results],
    titles=[f'Wiener Filter Kernel\nK={result["K"]}' for result in restored_results]
)

# show the restored image
show_images_project2(
    images=[result["restored_img"] for result in restored_results],
    titles=[f'Restored Image\n(Wiener Filter)\nK={result["K"]}' for result in restored_results]
)
```

### Use NLM (Non-Local Means) method to denoise Rician noise

Use the nlm function in package `skimage`.

```python
from skimage.restoration import denoise_nl_means, estimate_sigma

def applyNLM(img, patch_size: int = 7, patch_distance: int = 11, h: float = None, fast_mode: bool = True, dtype: np.dtype = np.float64):
    """ apply Non-Local Means (NLM) denoising to the image """
    # normalize the image
    img = imageNormalization(img, dtype=dtype)
    # estimate the noise standard deviation from the noisy image
    sigma_est = estimate_sigma(img, channel_axis=None)
    h = h if h is not None else 1.15 * sigma_est
    denoised_img = denoise_nl_means(
        img,
        patch_size=patch_size,
        patch_distance=patch_distance,
        h=h,
        fast_mode=fast_mode,
        channel_axis=None
    )
    return (denoised_img * 255).astype(dtype)

# patch_distance list
patch_distance_list = [1, 3, 5, 7, 11]
denoised_results = []

# apply NLM with different patch distances
for patch_distance in patch_distance_list:
    denoised_img = applyNLM(img_add_ricain, patch_distance=patch_distance, h=None)
    denoised_results.append({
        "patch_distance": patch_distance,
        "denoised_img": denoised_img
    })

# evaluate the denoised results
evaluation = evaluate_image(img, img_add_ricain)
print(f"Rician Noisy Image: MSE={evaluation['MSE']:.4f}, PSNR={evaluation['PSNR']:.4f} dB, SSIM={evaluation['SSIM']:.4f}")
for result in denoised_results: 
    evaluation = evaluate_image(img, result["denoised_img"])
    print(f"Patch Distance={result['patch_distance']}: MSE={evaluation['MSE']:.4f}, PSNR={evaluation['PSNR']:.4f} dB, SSIM={evaluation['SSIM']:.4f}")


# show the image
show_images_project2(
    images=[img, img_add_ricain] + [result["denoised_img"] for result in denoised_results],
    titles=['Original Image', 'Rician Noisy Image'] + [f'Denoised Image\n(Non-Local Means)\nPatch Distance={result["patch_distance"]}' for result in denoised_results]
)
```

Based on the results, the output of NLM is not good enough. The `MSE` and `PSNR` even turn to worse, but `SSIM` get better.

The result is funny, because when `patch_distance = 1`, the result is best.

However, the bad result is understandable. NLM is used to remove additive Gaussian Noise, while the Rician noise is not the Gaussian Noise but the Gaussian Noise in Spatial Domain and Frequency Domain. So in low frequence region, the denoise result is awful.
