```python
import numpy as np
```

```python
I = np.array([
    [50, 55, 60, 65, 70],
    [55, 60, 70, 75, 80],
    [60, 70, 150, 160, 90],
    [65, 75, 160, 170, 100],
    [70, 80, 90, 100, 110]
], dtype=np.float64)
```

```python
sobel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
])

sobel_y = np.array([
    [-1, -2, -1],
    [0, 0, 0],
    [1, 2, 1]
])
```

```python
def conv2(image, kernel, padding: str = 'zero'):
    # get busic information
    img_h, img_w = image.shape
    kernel_h, kernel_w = kernel.shape
    
    # 1. padding
    pad_h = (kernel_h - 1) // 2
    pad_w = (kernel_w - 1) // 2
    if padding == 'zero':
        padded_image = np.zeros((img_h + 2*pad_h, img_w + 2*pad_w), dtype=np.float64)
        padded_image[pad_h:pad_h+img_h, pad_w:pad_w+img_w] = image
    else:
        raise ValueError("目前仅支持 zero padding")

    # 2. Convolution
    result = np.zeros((img_h, img_w), dtype=np.float64)
    for i in range(img_h):
        for j in range(img_w):
            window = padded_image[i:i+kernel_h, j:j+kernel_w]
            result[i, j] = np.sum(window * kernel)
    
    return result
```

```python
def format_print(A):
    for row in A:
        formatted_row = " ".join([f"{val:8.3f}" for val in row])
        print(formatted_row)
```

```python
Gx = conv2(I, sobel_x, padding='zero')
Gy = conv2(I, sobel_y, padding='zero')
```

```python
Magnitude = np.sqrt(Gx**2 + Gy**2)
Direction = np.degrees(np.arctan2(Gy, Gx))
```

```python
print("Magnitude: ")
format_print(Magnitude)
```

```python
print("Direction: ")
format_print(Direction)
```
